```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report

import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

```python
N = 5000

# -----------------------------
# Demographics
# -----------------------------
age = np.random.randint(18, 75, N)

gender = np.random.choice(["Male", "Female"], size=N, p=[0.48, 0.52])
marital_status = np.random.choice(
    ["Single", "Married", "Divorced"], size=N, p=[0.45, 0.45, 0.10]
)

education_level = np.random.choice(
    ["High School", "Diploma", "Bachelor", "Postgraduate"],
    size=N,
    p=[0.25, 0.30, 0.30, 0.15]
)

employment_status = np.random.choice(
    ["Unemployed", "Employed", "Self-Employed"],
    size=N,
    p=[0.12, 0.70, 0.18]
)

# -----------------------------
# Income & Wealth
# -----------------------------
base_income = {
    "Unemployed": 4000,
    "Employed": 18000,
    "Self-Employed": 25000
}

monthly_income = np.array([
    np.random.lognormal(mean=np.log(base_income[e]), sigma=0.45)
    for e in employment_status
])

avg_monthly_balance = monthly_income * np.random.uniform(0.15, 0.45, N)

# -----------------------------
# Banking Relationship
# -----------------------------
account_tenure_years = np.clip(
    np.random.normal(loc=age / 10, scale=2),
    0.5,
    40
)

num_products = np.random.poisson(lam=2.2, size=N) + 1
num_products = np.clip(num_products, 1, 8)

# -----------------------------
# Credit & Risk
# -----------------------------
credit_score = (
    300
    + (monthly_income / np.max(monthly_income)) * 400
    + np.random.normal(0, 40, N)
)
credit_score = np.clip(credit_score, 300, 850)

has_credit_card = (credit_score > 520).astype(int)
has_loan = (credit_score > 560).astype(int)

loan_amount = np.where(
    has_loan == 1,
    monthly_income * np.random.uniform(4, 18, N),
    0
)

loan_interest_rate = np.where(
    has_loan == 1,
    np.random.uniform(0.08, 0.24, N),
    0
)

debt_to_income_ratio = np.where(
    has_loan == 1,
    loan_amount / (monthly_income * 12),
    0
)

# -----------------------------
# Engagement & Channels
# -----------------------------
digital_usage_score = (
    100
    - (age * 0.7)
    + (num_products * 6)
    + np.random.normal(0, 8, N)
)
digital_usage_score = np.clip(digital_usage_score, 5, 100)

branch_visits_per_year = np.clip(
    np.random.poisson(lam=(age / 15)), 0, 40
)

# -----------------------------
# Default Risk (Probabilistic)
# -----------------------------
prob_default = (
    0.45 * debt_to_income_ratio
    - 0.002 * credit_score
    + 0.000015 * loan_amount
    + np.random.normal(0, 0.02, N)
)

prob_default = np.clip(prob_default, 0.01, 0.85)

default_flag = (np.random.uniform(0, 1, N) < prob_default).astype(int)
```

```python

# Final Dataset

df = pd.DataFrame({
    "age": age,
    "gender": gender,
    "marital_status": marital_status,
    "education_level": education_level,
    "employment_status": employment_status,
    "monthly_income": monthly_income,
    "avg_monthly_balance": avg_monthly_balance,
    "account_tenure_years": account_tenure_years,
    "num_products": num_products,
    "credit_score": credit_score,
    "has_credit_card": has_credit_card,
    "has_loan": has_loan,
    "loan_amount": loan_amount,
    "loan_interest_rate": loan_interest_rate,
    "debt_to_income_ratio": debt_to_income_ratio,
    "digital_usage_score": digital_usage_score,
    "branch_visits_per_year": branch_visits_per_year,
    "prob_default": prob_default,
    "default_flag": default_flag
})
```
